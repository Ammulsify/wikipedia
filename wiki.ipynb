```python
import pandas as pd
import numpy as np

df = pd.read_csv('Data/merged_2026.csv')

# Simulate x positions based on month
month_order = ["January", "February", "March", "April", "May", "June"]
size_w = 780
padding = 0.5

# Calculate xScale positions (matching D3 scalePoint with padding 0.5)
n = len(month_order)
step = size_w / (n + padding * 2)
month_x = {m: step * (padding + i + 0.5) for i, m in enumerate(month_order)}

# Calculate yScale positions (log scale)
import math
size_h = 800
min_views = df['Views'].min()
max_views = df['Views'].max()

def log_scale(v):
    log_min = math.log10(min_views)
    log_max = math.log10(max_views)
    log_v = math.log10(v)
    return size_h - ((log_v - log_min) / (log_max - log_min)) * size_h

df['base_x'] = df['Month'].map(month_x)
df['base_y'] = df['Views'].apply(log_scale)

# Add jitter
np.random.seed(42)
df['jitter'] = np.random.uniform(-22, 22, len(df))
df['x'] = df['base_x'] + df['jitter']
df['y'] = df['base_y']

df.to_csv('Data/merged_2026.csv', index=False)
print("Done. Columns:", df.columns.tolist())
print(df[['Page', 'Month', 'x', 'y', 'jitter']].head())
```
